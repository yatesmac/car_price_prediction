# Data Preparation

```python
import pandas as pd
import numpy as np
```

```python
# Load the Data
data = '../data/external/final_scout_not_dummy.csv'
df = pd.read_csv(data)
df.head()
```

```python
# Removing duplicates
df = df.drop_duplicates()
df[df.duplicated()].sum()
```

```python
# Remove missing values
df = df.dropna()
df.isna().sum()
```

```python
# Format all text - removing spaces and hyphens.
df.columns = df.columns.str.lower().str.replace(' ', '_')
categorical = df.select_dtypes(include='object').columns.tolist()
for c in categorical:
    df[c] = df[c].str.lower().str.replace(' ', '_').replace('-', '_', regex=True)
```

```python
# Format all numerical values
numerical = df.select_dtypes('number').columns.tolist()
for n in numerical:
    df[n] = pd.to_numeric(df[n], errors='coerce')
```

```python
# Convert multi-label columns from strings to lists
multi_label = [
    'comfort_convenience',
    'entertainment_media',
    'extras',
    'safety_security']

for col in multi_label:
    df[col] = df[col].str.split(',')
```

```python
# Seperate categorical columns from multi-label columns 
categorical = list(set(categorical)-set(multi_label))

categorical, numerical, multi_label
```

```python
def remove_outliers(df, target):
    ''' Removing Outliers using Inter Quartile Range'''
    # Calculate the upper and lower limits
    Q1 = df[target].quantile(0.25)
    Q3 = df[target].quantile(0.75)
    IQR = Q3 - Q1
    lower_range = Q1 - 1.5 * IQR
    upper_range = Q3 + 1.5 * IQR

    # Removing the outliers
    df = df[~((df[target] < lower_range) | (df[target] > upper_range))]
    return df
```

```python
target = 'price'
df = remove_outliers(df, target)
```

## Splitting the dataset

```python
from sklearn.model_selection import train_test_split

# Split the dataframe into 60%, 20%, 20% parts
# Create Training, Validation and Testing datasets.

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=42)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=42)
```

## Saving Datasets

```python
# Saving datasets
train_dfs = [
    (df_full_train, 'df_full_train'), 
    (df_train, 'df_train'),
    (df_val, 'df_val')
]
test_dfs = [
    (df_test, 'df_test')
    ]

train_dir = '../data/train'
test_dir = '../data/test'

for df, df_name in train_dfs:
    df.to_csv(f'{train_dir}/{df_name}.csv', index=False)
    
for df, df_name in test_dfs:
    df.to_csv(f'{test_dir}/{df_name}.csv', index=False)
```
