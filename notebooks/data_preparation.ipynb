# Data Preparation

```python
import pandas as pd
import numpy as np
# import seaborn as sns
```

```python
# Load the Data
data = '../data/external/final_scout_not_dummy.csv'
df = pd.read_csv(data)
df.head()
```

```python
# Removing duplicates
df = df.drop_duplicates()
df[df.duplicated()].sum()
```

```python
# Remove missing values
df = df.dropna()
df.isna().sum()
```

```python
# Format all text - removing spaces and hyphens.
df.columns = df.columns.str.lower().str.replace(' ', '_')
categorical = df.select_dtypes(include='object').columns.tolist()
for c in categorical:
    df[c] = df[c].str.lower().str.replace(' ', '_').replace('-', '_', regex=True)
```

```python
# Format all numerical values
numerical = df.select_dtypes('number').columns.tolist()
for n in numerical:
    df[n] = pd.to_numeric(df[n], errors='coerce')
```

```python
# Convert multi-label columns from strings to lists
multi_label = [
    'comfort_convenience',
    'entertainment_media',
    'extras',
    'safety_security']

for col in multi_label:
:
    df[col] = df[col].str.split(',')
```

```python
# Seperate categorical columns from multi-label columns 
categorical = list(set(categorical)-set(multi_label))

categorical, numerical, multi_label
```

```python
df.head()
```

## Target Variable

```python
# The Target Variable
target = 'price'
# Preparing the target Variable:
# y_full is a df of the Target Variable
y_full = df[target]

# Remove Target from datase and column list
numerical.remove(target)
X_full = df.drop(columns=target)
```

```python
# Remove highly correlated variable from dataset and column list.
high_corr = ['hp_kw'] 
for h in high_corr:
    numerical.remove(h)
X_full = X_full.drop(columns=high_corr)
```

## Splitting the dataset

```python
from sklearn.model_selection import train_test_split

# Split the dataframe into 60%, 20%, 20% parts
# Create Training, Validation and Testing datasets.

X_full_train, X_test, y_full_train, y_test = train_test_split(X_full, y_full, test_size=0.2, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_full_train, y_full_train, test_size=0.25, random_state=42)
```

## Saving Datasets

```python
# Saving datasets
trains = [
    X_full_train, 
    y_full_train, 
    X_train, 
    y_train, 
    X_val, 
    y_val]
tests = [X_test, y_test]

train_dir = 'data/raw/train'
test_dir = 'data/raw/test'

for t in trains:
    t.to_csv(f'{train_dir}/{t}.csv', index=False)
    
for t in tests:
    t.to_csv(f'{test_dir}/{t}.csv', index=False)
```
